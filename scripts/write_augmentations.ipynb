```python
# Base flows

df = pd.read_csv("../isp_sheets_23/network_capability/flow_path_capability.csv")

# Add options

expand = pd.read_csv("../isp_sheets_23/network_capability/linearised_costs.csv",index_col=0)

# Committed and anticipated

# VIC-SESA Heywood 100
# SESA-CSA 100
# SNSW-CSA and VIC-SESA combined 650,800
# CNSW - SNW 1160,0 north south combined
# SNSW-CNSW VNI, 250,0
# CNSW-NNSW 0,300

# Pretend these have been added # new constraints just have an increase amount

# SNSW-CSA 800
# TAS-VIC Marinus Link
# SNSW–CNSW Humelink

# Add variables

n.model.add_variables()

# Costs must increase each year 

for _,row in df.iterrows():


    link = row["Plain Name"]


    # only add objective every year BRUH
    for year in n.investment_periods:

        # skip 2024
        if year==2024:
            continue

        # grow by 1% each year

        link_year = n.model.variables["Link-p"].loc[(year,),link]

        if link in expand.index:

            expand_row = expand.loc[link]

            if expand_row["Forward increase"] > 0:

                if year==2025:

                    increase_amount = n.model.add_variables(lower=0,upper=expand_row["Forward increase"],name=f"{link} + forward increase")
                
                else:

                    increase_amount = n.model.variables[f"{link} + forward increase"]

                n.model.add_constraints(link_year <= row["Max forward"] + increase_amount,
                                        name = link + f"-upper-{year}")
                
                n.model.objective += expand.loc[link,"Forward cost"]*(1.001**(year-2024))*increase_amount
                
            if expand_row["Reverse increase"] > 0:

                if year==2025:

                    increase_amount = n.model.add_variables(lower=0,upper=expand_row["Reverse increase"],name=f"{link} + reverse increase")
                
                else:

                    increase_amount = n.model.variables[f"{link} + reverse increase"]

                n.model.add_constraints(-link_year <= row["Max reverse"] + increase_amount,
                                        name = link + f"-lower-{year}")
                
                n.model.objective += expand.loc[link,"Reverse cost"]*(1.001**(year-2024))*increase_amount

```

```python
import geopandas as gpd
from shapely.geometry import Point
from collections import defaultdict as dd

def check_point_in_rez(lat, lon, rez_gdf):
    """
    Check if (lat, lon) point lies inside any REZ polygon
    """
    point = Point(lon, lat)  # shapely uses (x=lon, y=lat)
    
    matches = rez_gdf[rez_gdf.geometry.contains(point)]
    if matches.empty:
        # print(f"Point ({lat}, {lon}) is NOT inside any REZ zone.")

        return None
    else:
        # print(f"Point ({lat}, {lon}) IS inside the following REZ zone(s):")

        rez = matches['Name'].values[0].split(" ")
        # print(rez)

        return rez[0]

# Path to your AEMO KMZ file
# KMZ_FILE = "aemo-indicative-rez-boundaries-2025-gis-data.kmz"
# rez_gdf = load_rez_from_kmz(KMZ_FILE)
KML_FILE = "../data/indicative-rez-boundaries-2024-gis-data.kml"
rez_gdf = gpd.read_file(KML_FILE, driver='KML')

gen_rez_map = dd(dict)

df = pd.read_csv("../isp_sheets_23/summary_mapping/existing.csv")

for _,row in df.iterrows():

    # materially affects REZ transmission if VERY large (maybe not coal or gas or hydro if that's baseline?)

    # check PLEXOS

    gen = row["Existing generator"]

    fuel = row["Fuel type"]

    gen_lat,gen_lon = get_lat_lon(gen + " Australia", API_KEY)

    rez = check_point_in_rez(gen_lat, gen_lon, rez_gdf)

    if rez != None:
        print(rez)
        gen_rez_map[gen]["REZ"] = rez
        gen_rez_map[gen]["Fuel"] = fuel
```
